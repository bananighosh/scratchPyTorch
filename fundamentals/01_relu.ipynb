# 🟢 Easy: Implement ReLU

Implement the **ReLU** (Rectified Linear Unit) activation function from scratch.

$$\text{ReLU}(x) = \max(0, x)$$

### Signature
```python
def relu(x: torch.Tensor) -> torch.Tensor:
    ...
```

### Rules
- Do **NOT** use `torch.relu`, `F.relu`, `torch.clamp`, or any built-in activation
- Must support autograd (gradients should flow back)

### Example
```
Input:  tensor([-2., -1., 0., 1., 2.])
Output: tensor([ 0.,  0., 0., 1., 2.])
```

````python
# ✏️ YOUR IMPLEMENTATION HERE

def relu(x: torch.Tensor) -> torch.Tensor:
    # return torch.max(torch.zeros_like(x), x)
    # return torch.where(x > 0, x, torch.zeros_like(x))
    return x * (x > 0)
    # pass  # Replace this
````

````python
# 🧪 Test your implementation (feel free to add more debug prints)
x = torch.tensor([-2., -1., 0., 1., 2.])
print("Input: ", x)
print("Output:", relu(x))
print("Shape: ", relu(x).shape)
````

````python
## Benchmarking
device = "cuda" if torch.cuda.is_available() else "cpu"

x = torch.randn(10_000_000, device = device, requires_grad = True)

# Compilation and warm up
for _ in range(5):
  y = relu(x)

torch.cuda.synchronize()

start = torch.cuda.Event(enable_timing = True)
end = torch.cuda.Event(enable_timing = True)

start.record()

for _ in range(100):
  y = relu(x)

end.record()
torch.cuda.synchronize()

print(start.elapsed_time(end), "ms")
````

````python
#  CPU only benchmark
import torch
import torch.utils.benchmark as benchmark

x = torch.randn(10_000_000)

def relu(x):
    return x * (x > 0)

result = benchmark.Timer(
    stmt="relu(x)",
    globals={"relu": relu, "x": x},
).blocked_autorange(min_run_time=1)

print(result)
````
